# Partie D — Projet : le jeu de la vie

Le **jeu de la vie**, inventé par le mathématicien John Conway en 1970, est une grille de cellules **vivantes** (`1`) ou **mortes** (`0`). À chaque **génération**, toutes les cellules changent **en même temps** selon deux règles, qui dépendent du nombre de **voisines vivantes** (parmi les 8 cases autour) :

- une cellule **morte** avec **exactement 3** voisines vivantes **naît** ;
- une cellule **vivante** avec **2 ou 3** voisines vivantes **survit** ; sinon elle **meurt** (d'isolement ou d'étouffement).

Avec ces deux règles très simples, on voit apparaître des formes qui bougent, clignotent, se reproduisent... Les cases en dehors de la grille sont considérées comme mortes.
