# 1. Une matrice, c'est quoi ?

Une **matrice** est un tableau de nombres (ou d'autres valeurs) rangés en **lignes** et en **colonnes**, comme un tableau de notes :

| | Maths | NSI | Physique | Anglais |
|:---|:---:|:---:|:---:|:---:|
| **Élève 0** | 12 | 15 | 9 | 14 |
| **Élève 1** | 8 | 11 | 13 | 10 |
| **Élève 2** | 17 | 16 | 18 | 19 |

En Python, on la représente par un **tableau de tableaux** : un grand tableau dont chaque élément est **une ligne**.

````python
notes = [[12, 15, 9, 14],
         [8, 11, 13, 10],
         [17, 16, 18, 19]]

print(notes)
print(len(notes), "lignes")
print(len(notes[0]), "colonnes")
````

#### 💡 À retenir
- `notes` contient **3 éléments** : les 3 lignes. Donc `len(notes)` donne le **nombre de lignes**.
- Chaque ligne contient 4 notes. Donc `len(notes[0])` (la longueur de la première ligne) donne le **nombre de colonnes**.
- On écrit souvent la matrice **sur plusieurs lignes**, une ligne de code par ligne de la matrice : c'est plus lisible, et Python l'accepte tant qu'on est entre crochets.

On retrouve des matrices partout : un **plateau de jeu** (morpion, échecs, puissance 4), une **image** (une grille de pixels), un **tableau de notes**, une grille de **sudoku**...

# 2. Accéder à une case : `m[i][j]`

- `m[i]` est la **ligne** numéro `i` (c'est un tableau) ;
- `m[i][j]` est la case de la **ligne `i`** et de la **colonne `j`**.

Comme pour les listes, **on numérote à partir de 0**.

```
              j=0  j=1  j=2  j=3
    i=0  →  [ 12,  15,   9,  14 ]
    i=1  →  [  8,  11,  13,  10 ]
    i=2  →  [ 17,  16,  18,  19 ]
```

````python
print(notes[0])        # la ligne 0 (un tableau)
print(notes[2][3])     # ligne 2, colonne 3 : l'anglais de l'élève 2

ligne = notes[1]       # on peut aussi le faire en deux temps
print(ligne[2])        # c'est la même chose que notes[1][2]
print(notes[1][2])
````

#### 💡 À retenir
- `notes[1][2]` se lit **de gauche à droite** : d'abord on prend la ligne 1 (`notes[1]`), puis dans cette ligne la case 2.
- **Ligne d'abord, colonne ensuite.** Inverser les deux (`m[j][i]`) est l'erreur la plus fréquente.
- Si on sort du tableau (par exemple `notes[3][0]` alors qu'il n'y a que 3 lignes), Python s'arrête avec une erreur `IndexError`.

# 3. Modifier une case

On modifie une case comme dans une liste, avec `=` :

````python
notes[1][0] = 10       # l'élève 1 a maintenant 10 en maths
print(notes)
````

# 4. Parcourir une matrice

Pour passer sur **toutes** les cases, il faut **deux boucles imbriquées** : une pour les lignes, et dans chaque ligne une pour les colonnes.

## 4.1 Parcours par indices (quand on a besoin de la position)

````python
for i in range(len(notes)):              # i : numéro de ligne
    for j in range(len(notes[i])):       # j : numéro de colonne
        print("ligne", i, "colonne", j, ":", notes[i][j])
````

## 4.2 Parcours par valeurs (quand seule la valeur compte)

````python
for ligne in notes:          # ligne est un tableau
    for note in ligne:       # note est un nombre
        print(note, end=" ")
    print()                  # retour à la ligne à la fin de chaque ligne
````

#### 💡 À retenir
- La boucle **extérieure** parcourt les lignes, la boucle **intérieure** parcourt les cases d'une ligne. Pour une matrice de `n` lignes et `p` colonnes, la boucle intérieure tourne donc `n × p` fois en tout.
- **Par indices** (`for i in range(...)`) : indispensable dès qu'on veut **modifier** une case ou connaître sa **position**.
- **Par valeurs** (`for ligne in notes`) : plus court, parfait pour **lire** (somme, maximum, recherche...).
- `print(x, end=" ")` affiche `x` suivi d'un espace **sans aller à la ligne**.

# 5. Afficher une matrice proprement

`print(notes)` affiche tout sur une seule ligne, ce qui devient vite illisible. On écrit une fonction qui affiche une ligne de la matrice par ligne d'écran :

````python
def afficher(m):
    for ligne in m:
        for x in ligne:
            print(x, end="\t")   # \t : une tabulation, pour aligner les colonnes
        print()


afficher(notes)
````

# 6. Construire une matrice

## 6.1 Avec deux boucles

On fabrique chaque ligne, puis on l'ajoute à la matrice :

````python
def matrice(n, p, valeur):
    """Renvoie une matrice de n lignes et p colonnes remplie avec valeur."""
    m = []
    for i in range(n):
        ligne = []               # une NOUVELLE ligne à chaque tour
        for j in range(p):
            ligne.append(valeur)
        m.append(ligne)
    return m


afficher(matrice(2, 5, 0))
````

## 6.2 En compréhension

On peut écrire la même chose en une ligne, avec une liste en compréhension **dans** une liste en compréhension :

````python
zeros = [[0 for j in range(5)] for i in range(2)]
afficher(zeros)
print()

# La case (i, j) peut dépendre de i et de j : la table de multiplication
table = [[(i + 1) * (j + 1) for j in range(10)] for i in range(10)]
afficher(table)
````

## 6.3 ⚠️ Le piège classique : `[[0] * 3] * 3`

Pour une liste simple, `[0] * 3` donne bien `[0, 0, 0]`. On est donc tenté d'écrire `[[0] * 3] * 3` pour une matrice 3 × 3. Regardons ce qui se passe :

````python
mauvaise = [[0] * 3] * 3
print(mauvaise)              # ça a l'air correct...

mauvaise[0][0] = 5           # on modifie UNE case
print(mauvaise)              # ... et toute la colonne a changé !
````

**Pourquoi ?** `[0] * 3` crée **une** ligne. Puis `* 3` ne crée pas trois lignes : il met **trois fois la même ligne** dans la matrice. Les trois « lignes » sont en réalité **un seul et même tableau** ; si on le modifie par une case, on le voit modifié aux trois endroits.

On peut le vérifier avec `is`, qui teste si deux variables désignent **le même objet** :

````python
print(mauvaise[0] is mauvaise[1])     # True : c'est le même tableau

bonne = [[0] * 3 for i in range(3)]   # ici, une NOUVELLE ligne à chaque tour
bonne[0][0] = 5
print(bonne)
print(bonne[0] is bonne[1])           # False : trois tableaux différents
````

#### 💡 À retenir
- **À ne jamais écrire :** `[[0] * p] * n`.
- **À écrire :** deux boucles (6.1) ou `[[0] * p for i in range(n)]` (6.2) : la ligne est recréée à chaque tour.
- Pour **voir** ce qui se passe en mémoire, coller ce code sur [pythontutor.com](https://pythontutor.com) : on voit les trois flèches pointer vers le même tableau. C'est une question qui tombe souvent en contrôle.

# 7. Copier une matrice

Même piège avec `=` : `b = notes` ne copie pas la matrice, `b` est **un autre nom** pour le même tableau.

````python
b = notes
b[0][0] = 0
print(notes[0][0])           # 0 : notes a été modifiée aussi !
````

Pour obtenir une vraie copie, il faut recopier **chaque ligne** dans un nouveau tableau :

````python
def copie(m):
    res = []
    for ligne in m:
        nouvelle = []
        for x in ligne:
            nouvelle.append(x)
        res.append(nouvelle)
    return res


c = copie(notes)
c[0][0] = 20
print(notes[0][0], c[0][0])  # 0 20 : notes n'a pas bougé
````

#### 💡 À retenir
- `b = m` : deux noms, **une seule** matrice.
- Une vraie copie recrée **la matrice et chacune de ses lignes**. Recopier seulement la grande liste ne suffit pas : les lignes seraient encore partagées.
