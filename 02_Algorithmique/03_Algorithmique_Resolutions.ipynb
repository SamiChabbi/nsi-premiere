## Exercice 1

Pour chacune des boucles suivantes, dire si elle termine. Si oui, justifier en donnant un variant de boucle.

### Exercice 1 a)

#### 📝 ÉNONCÉ

```python
while i < 10:
    i = i + 2
```

#### ✅ RÉSOLUTION

**Elle termine.** Variant : **`10 - i`**.
1. C'est un entier (on suppose que `i` est un entier).
2. À chaque tour, `i` augmente de 2, donc `10 - i` **diminue de 2**.
3. Si `10 - i <= 0`, alors `i >= 10`, donc la condition `i < 10` est fausse : la boucle s'arrête.

````python
i = 1
while i < 10:
    print("i =", i, "   variant 10 - i =", 10 - i)
    i = i + 2
print("fin, i =", i)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Méthode : la condition est `i < 10`, c'est-à-dire « `10 - i` est positif ». On essaie donc `10 - i`.
- Ici `i` **augmente**, mais le variant, lui, **diminue** : le variant n'est pas forcément la variable elle-même.
- Même si `i` dépasse 10 sans tomber dessus (1, 3, 5, 7, 9, 11), ce n'est pas un problème : la condition est `<`, pas `!=`.

### Exercice 1 b)

#### 📝 ÉNONCÉ

```python
while n > 0:
    n = n / 2
```

#### ✅ RÉSOLUTION

**En théorie, elle ne termine pas** (pour `n > 0` au départ), et **il n'existe pas de variant**.
- `/` est la division **décimale** : `n` devient 8, 4, 2, 1, 0,5, 0,25, 0,125… Il reste toujours **strictement positif** : diviser par 2 un nombre positif ne donne jamais 0. La condition `n > 0` reste vraie pour toujours.
- `n` n'est pas un entier (dès qu'il passe sous 1), donc il ne peut pas servir de variant ; et aucune quantité entière ne diminue jusqu'à arrêter la boucle.

**En pratique, en Python, elle s'arrête quand même** : les nombres décimaux sont stockés avec une précision limitée, et quand `n` devient trop petit pour être représenté, Python l'arrondit à `0.0`. Cela arrive après environ 1 075 tours.

````python
n = 1.0
tours = 0
while n > 0:
    n = n / 2
    tours = tours + 1
print("Python s'est arrêté après", tours, "tours, avec n =", n)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- C'est la réponse attendue au sens du cours : **il n'y a pas de variant**, car `n` n'est pas entier et ne devient jamais négatif ni nul.
- L'arrêt en Python n'est **pas** une preuve de terminaison : il vient d'un défaut de la machine (la précision limitée des nombres décimaux, les « flottants »), pas de l'algorithme. Avec des nombres exacts, la boucle serait infinie.
- Comparer avec `n = n // 2` (division **entière**) : là, `n` devient 8, 4, 2, 1, 0 et la boucle termine, avec `n` comme variant.
- Si `n <= 0` au départ, on n'entre jamais dans la boucle.

### Exercice 1 c)

#### 📝 ÉNONCÉ

```python
while i <= 10 or i >= 100:
    i = i + 1
```

#### ✅ RÉSOLUTION

**Elle ne termine pas toujours**, donc on ne peut pas donner de variant valable pour toutes les valeurs de départ.

**Contre-exemple** : si `i = 100` au départ, la condition `i >= 100` est vraie ; `i` devient 101, 102, 103… et reste toujours `>= 100`. La boucle tourne à l'infini.

Pour les autres valeurs, elle termine :
- si `i <= 10` au départ, `i` augmente jusqu'à 11, où les deux conditions sont fausses ;
- si `11 <= i <= 99`, on n'entre même pas dans la boucle.

````python
def tours(i):
    securite = 0
    while (i <= 10 or i >= 100) and securite < 1000:
        i = i + 1
        securite = securite + 1
    return securite

print("départ i = 3   :", tours(3), "tours")
print("départ i = 50  :", tours(50), "tours")
print("départ i = 100 :", tours(100), "tours (arrêtée par le compteur de sécurité : infinie)")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Le `or` veut dire : on continue tant que **l'une des deux** conditions est vraie. Pour s'arrêter, il faut que **les deux** soient fausses, c'est-à-dire `11 <= i <= 99`.
- Quand `i >= 100`, augmenter `i` ne rapproche jamais de la zone d'arrêt : on s'en éloigne.
- Pour prouver qu'une boucle **ne termine pas**, il suffit de **un seul** contre-exemple.
- Si la précondition était « `i <= 10` », alors `11 - i` serait un variant.

### Exercice 1 d)

#### 📝 ÉNONCÉ

```python
while len(l) != 0:
    l.pop()
```

#### ✅ RÉSOLUTION

**Elle termine.** Variant : **`len(l)`**, la longueur de la liste.
1. C'est un entier.
2. `l.pop()` enlève le dernier élément : à chaque tour, `len(l)` **diminue de 1**.
3. Quand `len(l)` vaut 0, la condition `len(l) != 0` est fausse : la boucle s'arrête.

````python
l = [4, 7, 1]
while len(l) != 0:
    print("l =", l, "   variant len(l) =", len(l))
    l.pop()
print("fin, l =", l)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Ici, la condition est `!=`, d'habitude dangereuse (cours, partie 3). Mais `len(l)` diminue **de exactement 1** et ne peut **jamais être négatif** : il passe forcément par 0, donc il ne peut pas « sauter » la valeur d'arrêt.
- Comparer avec le cours : `while n != 0: n = n - 3` ne termine pas toujours, car `n` saute de 3 en 3 et peut passer à côté de 0.

````python
def pgcd(a, b):
    """Précondition : a et b sont des entiers strictement positifs."""
    while a != b:
        if a > b:
            a = a - b
        else:
            b = b - a
    return a
````

### Exercice 2 — question 1

#### 📝 ÉNONCÉ

Faire le suivi des valeurs de `a` et `b` pour `pgcd(12, 18)`.

#### ✅ RÉSOLUTION

| Tour | a (début) | b (début) | test `a > b` | action | a (fin) | b (fin) |
|:---|:---|:---|:---|:---|:---|:---|
| 1 | 12 | 18 | faux | `b = 18 - 12` | 12 | 6 |
| 2 | 12 | 6 | vrai | `a = 12 - 6` | 6 | 6 |
| — | 6 | 6 | `a != b` est faux : fin de la boucle | | | |

La fonction renvoie **6**.

````python
a = 12
b = 18
while a != b:
    if a > b:
        a = a - b
    else:
        b = b - a
    print("a =", a, "  b =", b)
print("pgcd(12, 18) =", pgcd(12, 18))
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Le PGCD est le plus grand diviseur commun : 6 divise bien 12 et 18, et c'est le plus grand.
- L'idée de l'algorithme : un diviseur commun de `a` et `b` divise aussi `a - b`. On remplace donc le plus grand par la différence, jusqu'à ce que les deux soient égaux.
- Suivi des valeurs = trace : une ligne par tour, et une ligne finale où la condition devient fausse.

### Exercice 2 — question 2

#### 📝 ÉNONCÉ

Que se passe-t-il pour `pgcd(5, 0)` ? Pourquoi la précondition exige-t-elle des valeurs strictement positives ?

#### ✅ RÉSOLUTION

Avec `a = 5` et `b = 0` : `a != b`, et `a > b`, donc `a = 5 - 0 = 5`. **Rien ne change** : `a` reste 5, `b` reste 0, et la boucle recommence la même chose **à l'infini**. La fonction ne renvoie jamais rien.

La précondition exige des valeurs **strictement positives** pour que **chaque tour fasse diminuer** quelque chose : quand on soustrait un nombre `>= 1`, la valeur baisse vraiment. Avec 0, on soustrait 0 et on tourne sur place. (Avec des nombres négatifs, la boucle peut aussi être infinie, par exemple `pgcd(-2, 3)` : `b` devient 5, 7, 9…)

````python
a = 5
b = 0
tours = 0
while a != b and tours < 5:      # compteur de sécurité : on montre seulement 5 tours
    if a > b:
        a = a - b
    else:
        b = b - a
    tours = tours + 1
    print("tour", tours, ": a =", a, "  b =", b)
print("... et ainsi de suite, sans fin")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- C'est le rôle d'une précondition (cours, partie 2) : si on appelle `pgcd(5, 0)`, on **ne respecte pas** la précondition, et la fonction ne promet plus rien, même pas de s'arrêter.
- La terminaison se démontre toujours **pour les entrées qui respectent la précondition** : la question 3 utilise justement « strictement positifs ».

### Exercice 2 — question 3

#### 📝 ÉNONCÉ

Proposer un variant de boucle pour prouver la terminaison de cet algorithme.

#### ✅ RÉSOLUTION

Variant : **`a + b`**.
1. C'est un **entier** (somme de deux entiers).
2. Il **diminue strictement** à chaque tour :
   - si `a > b`, on remplace `a` par `a - b` : la somme perd `b`, et `b >= 1` ;
   - sinon (`a < b`, car `a != b`), on remplace `b` par `b - a` : la somme perd `a`, et `a >= 1`.
   - Et `a`, `b` restent **strictement positifs** : quand `a > b`, `a - b >= 1` ; quand `b > a`, `b - a >= 1`. Donc la soustraction suivante enlèvera encore au moins 1.
3. `a + b` est un entier qui perd au moins 1 à chaque tour et qui reste toujours `>= 2` (car `a >= 1` et `b >= 1`). Il ne peut pas diminuer indéfiniment, donc la boucle s'arrête forcément.

````python
a = 12
b = 18
while a != b:
    print("a + b =", a + b)
    if a > b:
        a = a - b
    else:
        b = b - a
print("a + b =", a + b, " (fin)")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- `a` seul ou `b` seul ne marche pas : à certains tours, `a` ne change pas (c'est `b` qui diminue). Il faut une quantité qui diminue **à chaque** tour : la somme.
- Le point 3 est la forme du raisonnement : un entier qui diminue d'au moins 1 à chaque tour, et qui ne peut pas descendre en dessous d'une certaine valeur (ici 2), ne peut faire qu'un **nombre fini** de tours. On peut aussi prendre `a + b - 2` comme variant : il est `>= 0`, et la boucle s'arrête avant qu'il devienne négatif.
- On voit le lien avec la question 2 : avec `b = 0`, la somme ne diminue plus.

### Exercice 3 — question 1

#### 📝 ÉNONCÉ

Faire le suivi de la liste `l` pour `decompte([1,3,0,2])`.

#### ✅ RÉSOLUTION

À chaque tour, on regarde le **dernier** élément : s'il est positif, on lui enlève 1 ; s'il vaut 0, on le retire.

| Tour | l (début) | dernier élément | action | l (fin) |
|:---|:---|:---|:---|:---|
| 1 | [1, 3, 0, 2] | 2 > 0 | on enlève 1 | [1, 3, 0, 1] |
| 2 | [1, 3, 0, 1] | 1 > 0 | on enlève 1 | [1, 3, 0, 0] |
| 3 | [1, 3, 0, 0] | 0 | `pop` | [1, 3, 0] |
| 4 | [1, 3, 0] | 0 | `pop` | [1, 3] |
| 5 | [1, 3] | 3 > 0 | on enlève 1 | [1, 2] |
| 6 | [1, 2] | 2 > 0 | on enlève 1 | [1, 1] |
| 7 | [1, 1] | 1 > 0 | on enlève 1 | [1, 0] |
| 8 | [1, 0] | 0 | `pop` | [1] |
| 9 | [1] | 1 > 0 | on enlève 1 | [0] |
| 10 | [0] | 0 | `pop` | [] |

`len(l) > 0` est faux : fin, après **10 tours**. La liste est vide.

````python
l = [1, 3, 0, 2]
tour = 0
while len(l) > 0:
    if l[-1] > 0:
        l[-1] = l[-1] - 1
    else:
        l.pop()
    tour = tour + 1
    print("tour", tour, ":", l)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- La fonction ne renvoie rien : elle **modifie** la liste qu'on lui donne, jusqu'à la vider.
- Elle « décompte » le dernier nombre jusqu'à 0, le retire, puis passe au précédent.

### Exercice 3 — question 2

#### 📝 ÉNONCÉ

Proposer un variant de boucle pour prouver la terminaison de cet algorithme.

#### ✅ RÉSOLUTION

Variant : **la somme des éléments de `l` + la longueur de `l`**, c'est-à-dire `sum(l) + len(l)`.
1. C'est un **entier** (les éléments sont des entiers).
2. Il **diminue de 1 à chaque tour** :
   - si le dernier élément est `> 0`, on lui enlève 1 : la somme perd 1, la longueur ne change pas ;
   - sinon, on retire un 0 : la longueur perd 1, la somme ne change pas (on a retiré 0).
3. Il est toujours `>= 0` (les éléments sont positifs, d'après la précondition), et quand il vaut 0, la liste est vide (longueur 0), donc `len(l) > 0` est faux : la boucle s'arrête.

Pour `[1, 3, 0, 2]`, le variant vaut au départ `6 + 4 = 10` : c'est exactement le nombre de tours de la question 1.

````python
l = [1, 3, 0, 2]
while len(l) > 0:
    print(l, "   variant =", sum(l) + len(l))
    if l[-1] > 0:
        l[-1] = l[-1] - 1
    else:
        l.pop()
print(l, "   variant =", sum(l) + len(l), " (fin)")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Pourquoi pas `len(l)` seul ? Il ne diminue pas aux tours où on enlève 1 au dernier élément.
- Pourquoi pas `sum(l)` seul ? Il ne diminue pas aux tours où on retire un 0.
- Chaque tour fait **l'une ou l'autre** action. On additionne donc les deux quantités : à chaque tour, l'une des deux perd 1 et l'autre ne bouge pas. C'est la même astuce que l'exemple `(n - a) + (p - b)` du cours.
- La précondition « entiers **positifs** » sert ici : avec un nombre négatif à la fin, `l[-1] > 0` serait faux, on le retirerait, et la somme augmenterait.

### Exercice 4 — question 3

#### 📝 ÉNONCÉ

Corriger la fonction.

#### ✅ RÉSOLUTION

On initialise `m` avec **un élément de la liste**, le premier : `m = t[0]`. C'est possible car la précondition garantit que la liste n'est pas vide.

````python
def maximum(t):
    """Précondition : t est une liste non vide d'entiers.
    Postcondition : renvoie le plus grand élément de t."""
    m = t[0]
    for x in t:
        if x > m:
            m = x
    return m

print(maximum([3, 8, 2]))
print(maximum([-3, -1, -5]))
assert maximum([3, 8, 2]) == 8
assert maximum([-3, -1, -5]) == -1
assert maximum([7]) == 7
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Avec `m = t[0]`, `m` est **toujours un élément de la liste** : le résultat ne peut plus être une valeur inventée.
- C'est là que la précondition « non vide » sert : sans elle, `t[0]` provoquerait une erreur.
- C'est la version du cours (partie 1).

def puissance(x, n):

`x ** n` veut dire « `x` puissance `n` » : `x` multiplié par lui-même `n` fois (et `x ** 0` vaut 1).

````python
def puissance(x, n):
    """Précondition : n est un entier >= 0.
    Postcondition : renvoie x ** n."""
    r = 1
    k = 0
    while k < n:
        r = r * x
        k = k + 1
    return r
````

### Exercice 5 — question 1

#### 📝 ÉNONCÉ

Proposer un variant de boucle pour montrer que cet algorithme termine.

#### ✅ RÉSOLUTION

Variant : **`n - k`**.
1. C'est un **entier** (`n` est un entier par la précondition, `k` part de 0 et augmente de 1).
2. À chaque tour, `k` augmente de 1 et `n` ne change pas : `n - k` **diminue de 1**.
3. Si `n - k <= 0`, alors `k >= n`, donc la condition `k < n` est fausse : la boucle s'arrête.

````python
x = 2
n = 5
r = 1
k = 0
while k < n:
    print("k =", k, "  r =", r, "   variant n - k =", n - k)
    r = r * x
    k = k + 1
print("k =", k, "  r =", r, "   fin")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Même méthode qu'à l'exercice 1 a) : la condition est `k < n`, donc on essaie `n - k`.
- `n - k` est le **nombre de tours qu'il reste à faire**. C'est souvent une bonne façon de voir un variant.

### Exercice 5 — question 2

#### 📝 ÉNONCÉ

Montrer que `r == x ** k` est un invariant de la boucle.

#### ✅ RÉSOLUTION

**Idée :** `k` compte combien de fois on a multiplié `r` par `x`.

1. **Initialisation** : avant le premier tour, aucune multiplication : `k = 0` et `r = 1 = x ** 0`. L'invariant est vrai.

2. **Conservation** : si `r == x ** k` au début d'un tour, alors pendant le tour `r` est multiplié par `x` une fois de plus et `k` augmente de 1 pour le compter. Donc `r == x ** k` reste vrai à la fin du tour.

3. **Conclusion** : la boucle s'arrête quand `k = n`, donc `r == x ** n` : c'est la postcondition.

````python
x = 3
n = 4
r = 1
k = 0
print("avant la boucle : r =", r, "  x ** k =", x ** k, "  invariant :", r == x ** k)
while k < n:
    r = r * x
    k = k + 1
    print("fin du tour    : r =", r, "  x ** k =", x ** k, "  invariant :", r == x ** k)
print("puissance(3, 4) =", puissance(3, 4))
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- L'invariant dit en mots : « après `k` tours, `r` contient `x` multiplié `k` fois par lui-même ». C'est la réponse à la question « après `k` tours, que contient `r` ? » (méthode du cours).
- La **conservation** se rédige toujours pareil : « on suppose que c'est vrai au début du tour ; on regarde ce que fait chaque ligne ; c'est encore vrai à la fin ».
- La **conclusion** combine l'invariant avec la valeur de `k` à la sortie (`k = n`). Il faut justifier pourquoi `k` vaut **exactement** `n`, pas plus : `k` augmente de 1 en 1, donc il s'arrête dès qu'il atteint `n`.
- Le code ne prouve rien : il montre seulement l'invariant sur un exemple.

## Exercice 6

### Exercice 6 — question 1

#### 📝 ÉNONCÉ

Réécrire la fonction `maximum` (version corrigée) de l'exercice 4 en remplaçant la boucle `for` par une boucle `while`.

#### ✅ RÉSOLUTION

Au lieu de prendre les éléments un par un avec `for x in t`, on utilise un **indice** `i` qui part de 0 et augmente de 1 jusqu'à `len(t) - 1`. L'élément courant est `t[i]`.

````python
def maximum(t):
    """Précondition : t est une liste non vide d'entiers.
    Postcondition : renvoie le plus grand élément de t."""
    m = t[0]
    i = 0
    while i < len(t):
        if t[i] > m:
            m = t[i]
        i = i + 1
    return m

assert maximum([2, 5, 1, 4, 7]) == 7
assert maximum([-3, -1, -5]) == -1
assert maximum([7]) == 7
print("✔ Tous les tests passent")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Les trois éléments d'une boucle `while` qui remplace un `for` : **l'initialisation** de l'indice (`i = 0`), **la condition** (`i < len(t)`), **l'avancée** (`i = i + 1`). Oublier `i = i + 1` donne une boucle infinie.
- On commence à `i = 0` pour coller au tableau de la question 2 (qui commence à `i = 0`). Commencer à `i = 1` marcherait aussi, puisque `t[0]` est déjà dans `m`.
- Le cours conseille cette réécriture : avec un indice `i`, on peut écrire l'invariant en fonction de `i` (questions 3 et 4).

| i | t[i] | m avant | m après |
|:---|:---|:---|:---|
| 0 | 2 | 2 | 2 (2 > 2 est faux) |
| 1 | 5 | 2 | **5** (5 > 2) |
| 2 | 1 | 5 | 5 |
| 3 | 4 | 5 | 5 |
| 4 | 7 | 5 | **7** (7 > 5) |

Ensuite `i = 5`, et `5 < 5` est faux : fin. La fonction renvoie **7**.

````python
t = [2, 5, 1, 4, 7]
m = t[0]
i = 0
while i < len(t):
    avant = m
    if t[i] > m:
        m = t[i]
    print("i =", i, "  t[i] =", t[i], "  m avant =", avant, "  m après =", m)
    i = i + 1
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- « m avant » est la valeur au **début** du tour, « m après » à la **fin**. Le « m après » d'une ligne est le « m avant » de la ligne suivante.
- On remarque qu'à la fin de chaque ligne, `m` est le plus grand des éléments déjà regardés : 2, puis 5, 5, 5, 7. C'est l'invariant de la question 3.

### Exercice 6 — question 3

#### 📝 ÉNONCÉ

Proposer un invariant de boucle pour montrer que la fonction est correcte.

#### ✅ RÉSOLUTION

**Invariant** : « `m` est un élément de `t`, et `m` est supérieur ou égal aux éléments déjà parcourus (`t[0]` à `t[i - 1]`) ».

Autrement dit : **`m` est le plus grand des `i` premiers éléments** (ceux déjà parcourus).

#### 💡 EXPLICATION PÉDAGOGIQUE

- On l'obtient avec la méthode du cours : « après `i` tours, que contient `m` ? » → le plus grand des éléments déjà regardés. La trace de la question 2 le montre.
- On écrit les **deux parties** de la postcondition du cours (« le résultat appartient à `t` et est `>=` à tous ses éléments ») : c'est ce qui rend la conclusion facile à la question 4.
- La partie « `m` est un élément de `t` » est exactement ce qui manquait à la version fausse de l'exercice 4 (`m = 0`).

### Exercice 6 — question 4

#### 📝 ÉNONCÉ

Montrer que c'est bien un invariant.

#### ✅ RÉSOLUTION

1. **Initialisation** : avant le premier tour, `m = t[0]` et `i = 0`.
   - `m` est un élément de `t` (c'est `t[0]`) ;
   - « `m` est supérieur ou égal aux éléments déjà parcourus » : avec `i = 0`, **aucun** élément n'a encore été parcouru, il n'y a rien à vérifier, c'est vrai.

   (C'est la même façon de voir que dans l'exemple `somme` du cours, où la somme de 0 élément vaut 0.)

2. **Conservation** : au début du tour, `m` est un élément de `t` et `m` est supérieur ou égal aux éléments déjà parcourus. Pendant le tour, on compare `m` au nouvel élément `t[i]` : s'il est plus grand, `m` prend sa valeur. Dans les deux cas, `m` reste un élément de `t` et devient `>=` à `t[i]`. Il est donc encore supérieur ou égal à tous les éléments déjà parcourus, `t[i]` compris. L'invariant reste vrai.

3. **Conclusion** : la boucle s'arrête quand `i < len(t)` est faux, donc quand `i = len(t)` (`i` augmente de 1 en partant de 0). L'invariant dit alors : `m` est un élément de `t` et `m` est supérieur ou égal aux éléments déjà parcourus (`t[0]` à `t[len(t) - 1]`), c'est-à-dire **tous** les éléments. C'est la postcondition : `m` est le plus grand élément de `t`.

Et la boucle termine (variant `len(t) - i`), donc la fonction est **correcte**.

````python
t = [2, 5, 1, 4, 7]
m = t[0]
i = 0
while i < len(t):
    if t[i] > m:
        m = t[i]
    i = i + 1
    # vérification de l'invariant à la fin du tour
    deja_vus = []
    for k in range(i):
        deja_vus.append(t[k])
    print("i =", i, "  m =", m, "  max des", i, "premiers =", max(deja_vus), "  m dans t :", m in t)
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- C'est **exactement** le même plan que pour `somme` dans le cours et pour `puissance` à l'exercice 5 : initialisation, conservation (deux cas, à cause du `if`), conclusion avec `i = len(t)`.
- Quand il y a un `if` dans la boucle, la conservation se fait **cas par cas**.
- Le code ne fait que vérifier l'invariant sur un exemple ; la preuve, c'est le raisonnement.

### Exercice 7 a)

#### 📝 ÉNONCÉ

```python
def ajout_caractere(s: str, char: str):
  resultat = ""
  for c in s:
    resultat = resultat + c
  resultat = resultat + char
  return resultat
```

#### ✅ RÉSOLUTION

**`O(n)`**, avec `n` la longueur de `s` : complexité **linéaire**.

On compte :
- 1 affectation avant la boucle ;
- la boucle fait `n` tours (un par caractère de `s`), et chaque tour fait 1 concaténation et 1 affectation ;
- 2 opérations après la boucle (concaténation et affectation), puis le `return`.

Total : environ `2n + 4` opérations. On ignore le 2 et le 4 : **`O(n)`**.

````python
def ajout_caractere(s: str, char: str):
    resultat = ""
    for c in s:
        resultat = resultat + c
    resultat = resultat + char
    return resultat

print(ajout_caractere("bonjou", "r"))
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- La fonction recopie `s` caractère par caractère, puis ajoute `char` à la fin : elle renvoie `s + char`.
- Règle du cours : **une boucle** qui parcourt l'entrée, avec un travail constant à chaque tour → `O(n)`.
- `s: str` et `char: str` sont des **annotations** : elles indiquent le type attendu, mais ne changent rien à l'exécution.
- (Pour le prof : en réalité, Python recopie la chaîne à chaque `+`, ce qui rend chaque concaténation proportionnelle à la longueur. Le cours compte `a + b` comme une opération élémentaire ; avec cette convention, la réponse attendue est `O(n)`.)

### Exercice 7 b)

#### 📝 ÉNONCÉ

```python
def doublons(t):
    n = len(t)
    for i in range(n):
        for j in range(i + 1, n):
            if t[i] == t[j]:
                return True
    return False
```

#### ✅ RÉSOLUTION

**`O(n²)`** : complexité **quadratique**.

Le **pire des cas** est une liste **sans doublon** : la fonction ne s'arrête jamais en avance, elle fait toutes les comparaisons.
- pour `i = 0`, `j` va de 1 à `n - 1` : `n - 1` comparaisons ;
- pour `i = 1` : `n - 2` comparaisons ;
- … ;
- pour `i = n - 1` : 0 comparaison.

Total : `(n - 1) + (n - 2) + ... + 1 + 0 = n(n - 1) / 2 = n²/2 - n/2` comparaisons. On ne garde que le terme le plus grand, sans le nombre qui le multiplie : **`O(n²)`**.

````python
def comparaisons(n):
    t = []
    for k in range(n):
        t.append(k)            # pas de doublon : pire des cas
    compte = 0
    for i in range(n):
        for j in range(i + 1, n):
            compte = compte + 1
            if t[i] == t[j]:
                return compte
    return compte

for n in [10, 100, 1000]:
    print(n, "éléments :", comparaisons(n), "comparaisons")
````

#### 💡 EXPLICATION PÉDAGOGIQUE

- Règle du cours : **deux boucles imbriquées** sur la liste → `O(n²)`.
- La boucle intérieure commence à `i + 1` : elle fait de moins en moins de tours, ce qui divise le total par 2 environ. Mais un facteur 2 ne change pas l'ordre de grandeur : c'est toujours `O(n²)`.
- On voit sur les nombres : quand `n` est multiplié par 10, le nombre de comparaisons est multiplié par environ 100.
- Si un doublon est au début, la fonction s'arrête vite. Mais on regarde **le pire des cas**.

**`O(n²)`**, avec `n` la longueur de `email` : complexité **quadratique**.

- La boucle fait jusqu'à `n` tours. Le pire des cas est un email **sans `@`** (ou avec le `@` tout à la fin) : on parcourt tout.
